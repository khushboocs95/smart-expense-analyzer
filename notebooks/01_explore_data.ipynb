```python
import pandas as pd 
```

```python
df = pd.read_csv("../data/transactions.csv")
df.head()
df.isna().sum()
df.duplicated().sum()
print("Before removing Duplicate" , df.shape)
df.drop_duplicates(inplace=True)
print("After removing Duplicate" , df.shape)
df.isna().sum()
    
```

```python
median_amount = df["amount"].median()
print("Median amount is " , median_amount)
df["amount"] = df["amount"].fillna(median_amount)
df["amount"] = df["amount"].astype(int)
df.isna().sum()
```

```python
df[df["description"].str.islower()].head()
df["description"] = df["description"].str.strip().str.upper()
df[df["description"].str.islower()].shape
```

```python
df.to_csv("../data/transactions_clean.csv", index=False)
print("Saved" ,df.shape)
```
