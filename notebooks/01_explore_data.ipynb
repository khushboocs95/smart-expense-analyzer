```python
import pandas as pd 
```

```python
df = pd.read_csv("../data/transactions.csv")
df.head()
df.isna().sum()
df.duplicated().sum()
print("Before removing Duplicate" , df.shape)
df.drop_duplicates(inplace=True)
print("After removing Duplicate" , df.shape)
df.isna().sum()
    
```

```python
median_amount = df["amount"].median()
print("Median amount is " , median_amount)
df["amount"] = df["amount"].fillna(median_amount)
df["amount"] = df["amount"].astype(int)
df.isna().sum()
```

```python
df[df["description"].str.islower()].head()
df["description"] = df["description"].str.strip().str.upper()
df[df["description"].str.islower()].shape
```

```python
df.to_csv("../data/transactions_clean.csv", index=False)
print("Saved" ,df.shape)
```

```python
df["category"].value_counts()
df.groupby("category")["amount"].sum().sort_values(ascending=False)
df.groupby("category")["amount"].agg(["count" , "sum" , "mean"]).round(0)
df.groupby("payment_mode")["amount"].agg(["count" , "sum" , "mean"]).round(0)
```

## Insights
- Entertainment has the most transactions (413), but low total spend.
- Travel has the fewest transactions (373), but the highest total spend (₹11.4 lakh).
- Reason: Travel's average transaction (~₹3,000) is much higher, because of flight bookings.
